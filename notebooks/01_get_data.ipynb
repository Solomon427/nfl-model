```python
import pandas as pd
import numpy as np
from pathlib import Path
import nfl_data_py as nfl

RAW_DIR = Path("../data/raw")
PROCESSED_DIR = Path("../data/processed")

RAW_DIR.mkdir(parents=True, exist_ok=True)
PROCESSED_DIR.mkdir(parents=True, exist_ok=True)

YEARS = list(range(2011, 2021))
```

```python
draft = nfl.import_draft_picks(YEARS)

draft.head()
```

```python
draft.columns
```

```python
draft.to_csv(RAW_DIR / "draft_picks.csv", index=False)
```

```python
draft_values = nfl.import_draft_values()

draft_values.head()
```

```python
draft_values.columns
```

```python
draft_values.to_csv(RAW_DIR / "draft_values.csv", index=False)
```

```python
draft_av = draft[['pick', 'dr_av']]  # Options: 'w_av', 'car_av', 'dr_av'

draft_values_vs_av = draft_values.merge(
    draft_av,
    on="pick",
    how="left"
)

draft_values_vs_av.head()
```

```python
draft_values_vs_av.to_csv(RAW_DIR / "draft_values_vs_av.csv", index=False)
```
